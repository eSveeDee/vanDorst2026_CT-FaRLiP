# Photosynthetic Pigment Donor–Acceptor Distance Analysis

Computes pairwise inter-pigment distances from a PDB or mmCIF structure file, with a focus on donor→acceptor pairs relevant to energy transfer (e.g. phycobilisome bilin → PSI chlorophyll).

## Distance definitions

Three distance metrics are computed for each pigment pair:

- **Edge distance** — minimum atom-to-atom distance between the two pigments, restricted to conjugated-core atoms (see *Atom selection* below). This is the most conservative estimate of physical proximity and is used as the primary metric throughout.
- **Centroid distance** — Euclidean distance between the geometric centroids of the selected atom sets.
- **Representative distance** — distance between one representative point per pigment: the Mg atom for chlorophylls, the Fe atom for hemes, and the mean position of the four pyrrole nitrogens (NA–ND) for open-chain bilins (CYC and related). This approximates the position of the chromophore's coordination centre or the centre of the π-system.

## Atom selection

By default (`atom_selection_mode = "conjugated_core"`), only atoms belonging to the tetrapyrrole macrocycle or open-chain bilin conjugated system are included. Phytol tails and other substituents are excluded. This focuses distances on the chromophore region where electronic coupling is most relevant. The fallback for residue types without an explicit core definition is all non-hydrogen atoms.

Alternatively, `atom_selection_mode = "all_non_h"` uses all non-hydrogen atoms.

## Assumptions and limitations

- All calculations use a single model (first MODEL record in the file).
- Distances are purely geometric; no electronic coupling, spectral overlap, or orientation factors (κ²) are computed.
- The bilin pyrrole-nitrogen centroid is used as a proxy for the representative position; this does not account for chromophore conformation.
- Intra-chain and inter-chain pairs are not distinguished in the distance calculation itself, but are flagged in the output tables.

**Dependencies:** numpy, pandas, matplotlib, biopython, seaborn (optional)

## 1. Imports

```python
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.colors import to_rgba
from IPython.display import display
from Bio.PDB import PDBParser, MMCIFParser

try:
    import seaborn as sns
    HAVE_SEABORN = True
except ImportError:
    HAVE_SEABORN = False
```

## 2. User Parameters

```python
# ── Input / output ────────────────────────────────────────────────────────────

structure_path   = Path("supercomplex.pdb")         # Path to input PDB or mmCIF structure
out_dir_name     = "pigment_distance_output"   # Output directory name (created if absent)

# ── Pigment selection ─────────────────────────────────────────────────────────

# Residue names of donor pigments (energy source, e.g. PBS bilins)
focus_donor_resnames    = ["CYC"]

# Residue names of acceptor pigments (energy sink, e.g. PSI chlorophylls)
focus_acceptor_resnames = ["CLA", "F6C"]

# Residues to exclude from all calculations (e.g. non-photosynthetic cofactors)
excluded_resnames       = ["BCR", "HEM"]

# Restrict to specific chains (list of chain IDs), or None to include all
selected_chains         = None

# Atom selection: "conjugated_core" (tetrapyrrole core only) or "all_non_h"
atom_selection_mode     = "conjugated_core"

# ── Distance thresholds ───────────────────────────────────────────────────────

# Maximum edge distance (Å) to include a pair in the output table
contact_threshold_A     = 60.0

# Number of closest donor–acceptor pairs to display
top_n_pairs             = 50
```

## 3. Pigment Definitions and Helper Functions

Run once. Defines atom sets for each supported pigment type and the distance computation functions.

```python
# Supported photosynthetic residue names (auto-detected from structure)
SUPPORTED_RESNAMES = [
    "CLA", "CHL", "F6C", "BCL",          # chlorophylls
    "CYC", "PCB", "PEB", "PUB",          # bilins
    "PVB", "BVB", "DBV",
]

# Conjugated-core atom sets per residue type.
# None = fallback to all non-H atoms (used for open-chain bilins where
# no standardised core atom naming is available in the PDB).
TETRAPYRROLE_CORE = {
    "CLA": {"NA","NB","NC","ND","MG",
            "CHA","CHB","CHC","CHD",
            "C1A","C2A","C3A","C4A","CAA","CBA","CGA","O1A","O2A","CMA",
            "C1B","C2B","C3B","C4B","CAB","CBB","CMB",
            "C1C","C2C","C3C","C4C","CAC","CBC","CMC",
            "C1D","C2D","C3D","C4D","CAD","CBD","CGD","O1D","O2D","CMD","OBD"},
    "F6C": {"NA","NB","NC","ND","MG",
            "CHA","CHB","CHC","CHD",
            "C1A","C2A","C3A","C4A","CAA","CBA","CGA","O1A","O2A","CMA",
            "C1B","C2B","C3B","C4B","CAB","CBB","CMB","OMB",
            "C1C","C2C","C3C","C4C","CAC","CBC","CMC",
            "C1D","C2D","C3D","C4D","CAD","CBD","CGD","O1D","O2D","CMD","OBD"},
    "CHL": None,  # fallback to CLA core; handled below
    "BCL": None,
    "CYC": {"NA","NB","NC","ND",
            "CHA","CHB","CHD",
            "C1A","C2A","C3A","C4A","CAA","CBA","CGA","O1A","O2A","CMA",
            "C1B","C2B","C3B","C4B","CAB","CBB","CMB","OB",
            "C1C","C2C","C3C","C4C","CAC","CBC","CMC","OC",
            "C1D","C2D","C3D","C4D","CAD","CBD","CGD","O1D","O2D","CMD"},
    "PCB": None, "PEB": None, "PUB": None, "PVB": None, "BVB": None, "DBV": None,
}
# CHL shares the CLA core definition
TETRAPYRROLE_CORE["CHL"] = TETRAPYRROLE_CORE["CLA"]
TETRAPYRROLE_CORE["BCL"] = TETRAPYRROLE_CORE["CLA"]

# Representative atoms used for the 'representative_distance' metric
REP_ATOMS = {
    "CLA": ["MG"], "CHL": ["MG"], "F6C": ["MG"], "BCL": ["MG"],
    "CYC": ["NA","NB","NC","ND"],
    "PCB": ["NA","NB","NC","ND"], "PEB": ["NA","NB","NC","ND"],
    "PUB": ["NA","NB","NC","ND"], "PVB": ["NA","NB","NC","ND"],
    "BVB": ["NA","NB","NC","ND"], "DBV": ["NA","NB","NC","ND"],
}
```

```python
def residue_label(chain, residue):
    icode = residue.id[2].strip() if isinstance(residue.id[2], str) else ""
    return f"{chain.id}:{residue.resname}{residue.id[1]}{icode}"


def get_atom_coords(residue, mode="conjugated_core"):
    """Return (coords, atom_names) arrays for selected atoms in a residue."""
    allowed = TETRAPYRROLE_CORE.get(residue.resname) if mode == "conjugated_core" else None
    coords, names = [], []
    for atom in residue:
        name = atom.name.strip()
        if (atom.element or "").strip().upper() == "H" or name.startswith("H"):
            continue
        if mode == "all_non_h" or allowed is None or name in allowed:
            coords.append(atom.coord.astype(float))
            names.append(name)
    return np.asarray(coords, dtype=float), names


def get_representative_coord(residue, coords, names):
    """Return a single representative coordinate for a pigment (see header for definition)."""
    preferred = REP_ATOMS.get(residue.resname, [])
    matching = [xyz for xyz, n in zip(coords, names) if n in preferred]
    if matching:
        return np.mean(matching, axis=0)
    return coords.mean(axis=0)


def collect_pigments(structure, selected_resnames, mode="conjugated_core", chains=None):
    """Extract all selected pigments from structure and return as a list of dicts."""
    chain_filter = set(chains) if chains else None
    pigments = []
    idx = 0
    for model in structure:
        for chain in model:
            if chain_filter and chain.id not in chain_filter:
                continue
            for residue in chain:
                if residue.resname not in selected_resnames:
                    continue
                coords, atom_names = get_atom_coords(residue, mode)
                if len(coords) == 0:
                    continue
                pigments.append({
                    "index": idx,
                    "chain": chain.id,
                    "resname": residue.resname,
                    "resseq": residue.id[1],
                    "label": residue_label(chain, residue),
                    "coords": coords,
                    "atom_names": atom_names,
                    "centroid": coords.mean(axis=0),
                    "representative_coord": get_representative_coord(residue, coords, atom_names),
                })
                idx += 1
        break  # first model only
    return pigments


def compute_pairwise_table(pigments, threshold_A=None):
    """Compute all pairwise distances between pigments.
    
    Returns a long-format DataFrame with edge, centroid, and representative distances.
    Optionally filtered to pairs with edge_distance_A <= threshold_A.
    """
    records = []
    for i in range(len(pigments)):
        for j in range(i + 1, len(pigments)):
            pi, pj = pigments[i], pigments[j]
            # edge distance: minimum over all atom-atom pairs
            D = np.sqrt(((pi["coords"][:, None, :] - pj["coords"][None, :, :]) ** 2).sum(axis=2))
            ai, aj = np.unravel_index(np.argmin(D), D.shape)
            edge_d = float(D[ai, aj])
            if threshold_A is not None and edge_d > threshold_A:
                continue
            records.append({
                "i": i, "j": j,
                "label_i": pi["label"], "label_j": pj["label"],
                "chain_i": pi["chain"], "chain_j": pj["chain"],
                "resname_i": pi["resname"], "resname_j": pj["resname"],
                "resseq_i": pi["resseq"], "resseq_j": pj["resseq"],
                "same_chain": pi["chain"] == pj["chain"],
                "edge_distance_A": edge_d,
                "centroid_distance_A": float(np.linalg.norm(pi["centroid"] - pj["centroid"])),
                "representative_distance_A": float(np.linalg.norm(
                    pi["representative_coord"] - pj["representative_coord"])),
                "closest_atom_i": pi["atom_names"][ai],
                "closest_atom_j": pj["atom_names"][aj],
            })
    return pd.DataFrame(records).sort_values("edge_distance_A", ignore_index=True)


def focus_pairs(pair_df, donor_resnames, acceptor_resnames):
    """Extract and normalise donor→acceptor pairs from the full pairwise table."""
    d, a = set(donor_resnames), set(acceptor_resnames)
    fwd = pair_df[pair_df["resname_i"].isin(d) & pair_df["resname_j"].isin(a)].copy()
    rev = pair_df[pair_df["resname_j"].isin(d) & pair_df["resname_i"].isin(a)].copy()
    fwd = fwd.rename(columns={"label_i":"donor_label","label_j":"acceptor_label",
                               "chain_i":"donor_chain","chain_j":"acceptor_chain",
                               "resname_i":"donor_resname","resname_j":"acceptor_resname",
                               "resseq_i":"donor_resseq","resseq_j":"acceptor_resseq",
                               "closest_atom_i":"donor_atom","closest_atom_j":"acceptor_atom"})
    rev = rev.rename(columns={"label_j":"donor_label","label_i":"acceptor_label",
                               "chain_j":"donor_chain","chain_i":"acceptor_chain",
                               "resname_j":"donor_resname","resname_i":"acceptor_resname",
                               "resseq_j":"donor_resseq","resseq_i":"acceptor_resseq",
                               "closest_atom_j":"donor_atom","closest_atom_i":"acceptor_atom"})
    cols = ["donor_label","acceptor_label","donor_chain","acceptor_chain",
            "donor_resname","acceptor_resname","donor_resseq","acceptor_resseq",
            "donor_atom","acceptor_atom","same_chain",
            "edge_distance_A","centroid_distance_A","representative_distance_A"]
    return pd.concat([fwd[cols], rev[cols]], ignore_index=True).sort_values(
        "edge_distance_A", ignore_index=True)
```

## 4. Load Structure and Detect Pigments

```python
out_dir = Path(out_dir_name)
out_dir.mkdir(parents=True, exist_ok=True)

suffix = structure_path.suffix.lower()
if suffix == ".cif":
    parser = MMCIFParser(QUIET=True)
elif suffix == ".pdb":
    parser = PDBParser(QUIET=True)
else:
    raise ValueError(f"Unsupported format: {suffix}. Use .pdb or .cif")

structure = parser.get_structure(structure_path.stem, str(structure_path))

# Auto-detect photosynthetic pigments present in the structure
present = {r.resname for model in structure for chain in model
           for r in chain if str(r.id[0]).startswith("H_")}
selected_resnames = [r for r in SUPPORTED_RESNAMES
                     if r in present and r not in excluded_resnames]

counts = Counter(r.resname for model in structure for chain in model
                 for r in chain if r.resname in selected_resnames)
print("Detected pigments:")
display(pd.DataFrame(list(counts.items()), columns=["resname", "count"]).sort_values("count", ascending=False))
print(f"\nDonors:    {focus_donor_resnames}")
print(f"Acceptors: {focus_acceptor_resnames}")
```

## 5. Compute Pairwise Distances

```python
pigments = collect_pigments(structure, selected_resnames,
                            mode=atom_selection_mode, chains=selected_chains)
print(f"Total pigments collected: {len(pigments)}")

pair_df  = compute_pairwise_table(pigments, threshold_A=contact_threshold_A)
focus_df = focus_pairs(pair_df, focus_donor_resnames, focus_acceptor_resnames)

# Save
pair_df.to_csv(out_dir / "all_pairs.csv", index=False)
focus_df.to_csv(out_dir / "donor_acceptor_pairs.csv", index=False)
print(f"Saved tables to {out_dir.resolve()}")
```

## 6. Donor–Acceptor Distance Table

Rows are sorted by edge distance. Acceptor entries are colour-coded by chain to allow rapid identification of which chlorophylls belong to the same PSI subunit.

```python
# Assign a colour to each unique acceptor chain
acceptor_chains = sorted(focus_df["acceptor_chain"].unique())
chain_palette   = plt.cm.tab20.colors
chain_color_map = {ch: chain_palette[i % len(chain_palette)] for i, ch in enumerate(acceptor_chains)}

def color_acceptor_chain(val):
    """Background colour for acceptor_chain column cells."""
    rgba = to_rgba(chain_color_map.get(val, "white"), alpha=0.35)
    r, g, b, a = [int(x * 255) for x in rgba]
    return f"background-color: rgba({r},{g},{b},{a/255:.2f})"

def highlight_chlf(row):
    """Highlight entire row yellow if acceptor is Chl-f (F6C)."""
    return ["background-color: #f5c242" if row["acceptor_resname"] == "F6C" else ""
            for _ in row]

display_df = focus_df.head(top_n_pairs)[[
    "donor_label", "acceptor_label",
    "donor_chain", "acceptor_chain",
    "donor_resname", "acceptor_resname",
    "same_chain",
    "edge_distance_A", "centroid_distance_A", "representative_distance_A",
    "donor_atom", "acceptor_atom",
]].copy()

styled = (
    display_df.style
    .apply(highlight_chlf, axis=1)
    .applymap(color_acceptor_chain, subset=["acceptor_chain"])  # chain colour wins on that column
    .format({
        "edge_distance_A": "{:.2f}",
        "centroid_distance_A": "{:.2f}",
        "representative_distance_A": "{:.2f}",
    })
    .set_caption(
        f"Top {top_n_pairs} donor–acceptor pairs by edge distance. "
        "Acceptor chain colour indicates subunit identity. "
        "Yellow rows: Chl-f (F6C) acceptors."
    )
)

display(styled)

# Chain legend
patches = [mpatches.Patch(color=chain_color_map[ch], label=f"Chain {ch}")
           for ch in acceptor_chains]
patches.append(mpatches.Patch(color="#f5c242", label="Chl-f (F6C)"))
plt.figure(figsize=(max(3, (len(acceptor_chains) + 1) * 0.8), 0.6))
plt.legend(handles=patches, ncol=len(acceptor_chains) + 1, frameon=False,
           loc="center", title="Acceptor chain / pigment type")
plt.axis("off")
plt.tight_layout()
plt.show()
```

## 7. Edge Distance Distribution

```python
fig, ax = plt.subplots(figsize=(7, 4))

for resname, grp in focus_df.groupby("acceptor_resname"):
    ax.hist(grp["edge_distance_A"], bins=30, alpha=0.5, label=f"acceptor: {resname}")

ax.set_xlabel("Donor–acceptor edge distance (Å)")
ax.set_ylabel("Count")
ax.set_title("Distribution of donor–acceptor distances")
ax.legend(frameon=False)
plt.tight_layout()
plt.savefig(out_dir / "donor_acceptor_distance_distribution.pdf", bbox_inches="tight")
plt.show()
```
